```python
# Greedy Multi-Armed Bandit using Arithmetic Mean

n = int(input("Enter number of machines: "))

machines = []
q_values = {}

for i in range(n):
    name = input("\nEnter machine name: ")
    rewards = list(map(int, input("Enter rewards (space separated): ").split()))

    machines.append((name, rewards))
    q = sum(rewards) / len(rewards)
    q_values[name] = q

# Display results
print("\nEstimated Values:")
for name, rewards in machines:
    print(f"\nMachine {name}")
    print("Rewards:", rewards)
    print(f"Q({name}) = {q_values[name]:.2f}")

# Greedy selection
best = max(q_values, key=q_values.get)
print(f"\nGreedy Selected Machine: {best}")
```

```python
# Greedy Multi-Armed Bandit using Incremental Update

n = int(input("Enter number of machines: "))

Q = {}
N = {}
reward_data = {}

for i in range(n):
    name = input("\nEnter machine name: ")
    rewards = list(map(int, input("Enter rewards (space separated): ").split()))

    reward_data[name] = rewards
    Q[name] = 0.0
    N[name] = 0

# Update after every reward
for name in reward_data:
    print(f"\nMachine {name}")
    for r in reward_data[name]:
        N[name] += 1
        Q[name] = Q[name] + (r - Q[name]) / N[name]

        print(f"Reward = {r}, N = {N[name]}, Q({name}) = {Q[name]:.2f}")

# Display final values
print("\nFinal Estimated Values:")
for name in reward_data:
    print(f"Q({name}) = {Q[name]:.2f}")

# Greedy selection
best = max(Q, key=Q.get)
print(f"\nGreedy Selected Machine: {best}")
```
