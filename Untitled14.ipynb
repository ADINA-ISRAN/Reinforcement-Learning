**ε-Greedy Strategy for Multi-Armed Bandit**

```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays and epsilon
N_play = 10
epsilon = 0.2

# Part B – Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}       # Participation count
R = {'A': [], 'B': [], 'C': []}    # Rewards
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Q-values

total_reward = 0

# Part C – ε-Greedy Decision
for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        # Exploration
        selected = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        # Random choice among machines having the same maximum Q
        best_machines = [m for m in machines if Q[m] == max_q]
        selected = random.choice(best_machines)
        decision = "Exploitation"

    print("\n--------------------------------")
    print("Play:", play)
    print("Random number:", round(r, 3))
    print("Decision:", decision)
    print("Play with machine", selected)

    # Part D – Reward Collection
    reward = float(input("Enter reward: "))

    # Part E – Q-Value Update
    N[selected] += 1
    R[selected].append(reward)

    Q[selected] = Q[selected] + \
                  (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Updated N:", N)
    print("Updated Q:", {m: round(Q[m], 2) for m in machines})


# Part F – Display Final Results
print("\n================================")
print("FINAL RESULTS")
print("================================")

print("\nParticipation Count (N):")
for m in machines:
    print("Machine", m, ":", N[m])

print("\nRewards (R):")
for m in machines:
    print("Machine", m, ":", R[m])

print("\nQ-Values:")
for m in machines:
    print("Q(" + m + ") =", round(Q[m], 2))

print("\nTotal Reward after 10 plays:", total_reward)

# Best machine
best_machine = max(Q, key=Q.get)
print("Best Machine:", best_machine)
```
