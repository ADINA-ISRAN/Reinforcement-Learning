**Assignment1: Write Python code for policy iterative algorithm in a Deterministic Grid World
Objective**

```python
# Policy Iteration in a Deterministic 2x2 Grid World
# Grid:
# S0 | WALL
# S1 |  G

states = ["S0", "S1", "G"]
non_terminal_states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

gamma = 0.9
theta = 1e-8
normal_reward = -1
wall_reward = -1
goal_reward = 100

# Each entry is: action -> (next_state, reward)
transitions = {
    "S0": {
        "Up":    ("S0", wall_reward),
        "Down":  ("S1", normal_reward),
        "Left":  ("S0", wall_reward),
        "Right": ("S0", wall_reward)  # wall
    },
    "S1": {
        "Up":    ("S0", normal_reward),
        "Down":  ("S1", wall_reward),
        "Left":  ("S1", wall_reward),
        "Right": ("G", goal_reward)
    }
}

print("GRID WORLD")
print("S0 | WALL")
print("S1 |  G")
print("\\nTRANSITION INFORMATION")
for state in non_terminal_states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(f"{state} --{action} / reward {reward}--> {next_state}")

# Initial policy and values
policy = {"S0": "Up", "S1": "Up"}
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}
print("\\nInitial policy:", policy)
print("Initial values:", V)

# Policy Iteration
policy_iteration = 0
while True:
    policy_iteration += 1
    print(f"\\n========== POLICY ITERATION {policy_iteration} ==========")
    print("Policy being evaluated:", policy)

    # Policy Evaluation
    sweep = 0
    while True:
        sweep += 1
        old_values = V.copy()

        for state in non_terminal_states:
            action = policy[state]
            next_state, reward = transitions[state][action]
            V[state] = reward + gamma * old_values[next_state]

        V["G"] = 0.0
        delta = max(abs(V[s] - old_values[s]) for s in non_terminal_states)
        print(f"Sweep {sweep}: V(S0)={V['S0']:.6f}, "
              f"V(S1)={V['S1']:.6f}, V(G)=0.000000, delta={delta:.8f}")
        if delta < theta:
            break

    # Calculate Q values for all actions
    Q = {}
    for state in non_terminal_states:
        Q[state] = {}
        for action in actions:
            next_state, reward = transitions[state][action]
            Q[state][action] = reward + gamma * V[next_state]

    print("\\nAction-value table Q_pi(s,a)")
    print(f"{'State':<8}" + "".join(f"{a:<12}" for a in actions))
    for state in non_terminal_states:
        print(f"{state:<8}" + "".join(f"{Q[state][a]:<12.4f}" for a in actions))

    # Policy Improvement
    improved_policy = {}
    for state in non_terminal_states:
        improved_policy[state] = max(actions, key=lambda a: Q[state][a])

    print("\\nImproved policy:", improved_policy)

    # Policy stability
    if improved_policy == policy:
        print("Policy is stable. Stopping.")
        policy = improved_policy
        break
    policy = improved_policy

# Final Q* table from converged values
Q_star = {}
for state in non_terminal_states:
    Q_star[state] = {}
    for action in actions:
        next_state, reward = transitions[state][action]
        Q_star[state][action] = reward + gamma * V[next_state]

print("\\n========== FINAL OPTIMAL RESULTS ==========")
print(f"{'State':<8}{'Optimal Value':<18}{'Optimal Action':<16}")
for state in non_terminal_states:
    print(f"{state:<8}{V[state]:<18.4f}{policy[state]:<16}")
print(f"{'G':<8}{V['G']:<18.4f}{'Terminal':<16}")

print("\\nFinal optimal action-value table Q*(s,a)")
print(f"{'State':<8}" + "".join(f"{a:<12}" for a in actions))
for state in non_terminal_states:
    print(f"{state:<8}" + "".join(f"{Q_star[state][a]:<12.4f}" for a in actions))

print("\\nBellman optimality verification")
for state in non_terminal_states:
    max_q = max(Q_star[state].values())
    print(f"V*({state})={V[state]:.4f}, max_a Q*({state},a)={max_q:.4f}")

print("\\nOptimal path: S0 -> S1 -> G")
```
