**Assignment1: Write Python code for Value iterative algorithm in a Deterministic Grid World
Objective**

```python
# Value Iteration in a Deterministic 2x2 Grid World

# Define states
states = ['S0', 'S1', 'G']

# Define actions
actions = ['Up', 'Down', 'Left', 'Right']

# Discount factor
gamma = 0.9

# Convergence threshold
theta = 0.001

# Define transitions
# (state, action): (next_state, reward)

transitions = {
    ('S0', 'Up'): ('S0', -1),
    ('S0', 'Down'): ('S1', -1),
    ('S0', 'Left'): ('S0', -1),
    ('S0', 'Right'): ('S0', -1),

    ('S1', 'Up'): ('S0', -1),
    ('S1', 'Down'): ('S1', -1),
    ('S1', 'Left'): ('S1', -1),
    ('S1', 'Right'): ('G', 100)
}

# Initialize state values
V = {'S0': 0, 'S1': 0, 'G': 0}

# Value Iteration
iteration = 0

while True:
    delta = 0
    new_V = V.copy()

    for state in ['S0', 'S1']:
        q_values = {}

        for action in actions:
            next_state, reward = transitions[(state, action)]

            q_value = reward + gamma * V[next_state]
            q_values[action] = q_value

        new_V[state] = max(q_values.values())

        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print(f"\nIteration {iteration}")
    print("S0 =", round(V['S0'], 4))
    print("S1 =", round(V['S1'], 4))

    if delta < theta:
        break

# Calculate final Q-values
print("\nFINAL ACTION VALUES")

policy = {}

for state in ['S0', 'S1']:
    print(f"\nState: {state}")

    q_values = {}

    for action in actions:
        next_state, reward = transitions[(state, action)]

        q_value = reward + gamma * V[next_state]
        q_values[action] = q_value

        print(f"{action}: {q_value:.4f}")

    best_action = max(q_values, key=q_values.get)
    policy[state] = best_action

# Display final results
print("\nOPTIMAL STATE VALUES")

for state in states:
    print(f"V*({state}) = {V[state]:.4f}")

print("\nOPTIMAL POLICY")

for state, action in policy.items():
    print(f"{state} -> {action}")

print("\nOptimal Path: S0 -> S1 -> G")
```
