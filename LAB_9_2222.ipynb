```python
gamma = 0.9

# Bellman equations:
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

# Solve algebraically:
# V0 = -1 + 0.81 * (89.9 + 0.09 * V0)
# 0.9271 * V0 = 71.819

V0 = 71.819 / 0.9271
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_down = -1 + gamma * V1
Q0_right = -1

Q1_right = 100
Q1_up = -1 + gamma * V0

# Verify Bellman expectation equations
V0_check = 0.9 * Q0_down + 0.1 * Q0_right
V1_check = 0.9 * Q1_right + 0.1 * Q1_up

print("ASSIGNMENT 1 RESULTS")
print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")

print(f"Q(S0, Down) = {Q0_down:.2f}")
print(f"Q(S0, Right) = {Q0_right:.2f}")
print(f"Q(S1, Right) = {Q1_right:.2f}")
print(f"Q(S1, Up) = {Q1_up:.2f}")

print("\nVERIFICATION")
print(f"V(S0) from Q values = {V0_check:.2f}")
print(f"V(S1) from Q values = {V1_check:.2f}")
```
